# Question 1: What is Computational Linguistics and how does it relate to NLP?

**Answer:**

Computational Linguistics is an interdisciplinary field that focuses on the computational modeling and analysis of natural language using computer algorithms. It combines knowledge from linguistics, computer science, artificial intelligence, and cognitive science to enable machines to understand and process human language.

The goal of computational linguistics is to design systems that can process language data, such as speech recognition systems, grammar checkers, and machine translation tools.

Natural Language Processing (NLP) is a practical application of computational linguistics. While computational linguistics focuses on theoretical models of language processing, NLP uses these models to build real-world applications such as chatbots, sentiment analysis systems, and translation tools.

Thus, computational linguistics provides the theoretical foundation, while NLP focuses on implementing these theories in practical systems.

# Question 2: Briefly describe the historical evolution of Natural Language Processing.

**Answer:**

The development of Natural Language Processing has evolved through several stages.

**1. Rule-Based Systems (1950s–1980s)**  
Early NLP systems relied on manually written grammar rules and linguistic knowledge. These systems attempted to process language through predefined rules and dictionaries.

**2. Statistical NLP (1990s–2000s)**  
With the availability of large datasets and improved computing power, statistical methods such as n-grams and Hidden Markov Models were introduced. These methods learned patterns from data rather than relying entirely on rules.

**3. Machine Learning Era (2000s–2015)**  
Machine learning algorithms such as Support Vector Machines and Naive Bayes were widely used for tasks like text classification, named entity recognition, and sentiment analysis.

**4. Deep Learning Era (2015–Present)**  
Recent NLP systems use deep learning techniques such as neural networks and transformer architectures. Modern models like BERT and GPT can understand and generate human language with high accuracy.

# Question 4: What is text normalization and why is it essential in text processing tasks?

**Answer:**

Text normalization is the process of converting raw text into a standardized format that can be easily processed by machines.

Common normalization techniques include converting text to lowercase, removing punctuation, removing stopwords, correcting spelling, and performing stemming or lemmatization.

Normalization is essential because natural language data often contains inconsistencies such as capitalization differences, abbreviations, and noisy characters. By standardizing the text, NLP models can process data more effectively and achieve better accuracy.

# Question 5: Compare and contrast stemming and lemmatization with suitable examples.

**Answer:**

Stemming and lemmatization are techniques used to reduce words to their root or base form.

**Stemming** removes suffixes from words using simple rules. For example, the word "running" may be reduced to "run".

**Lemmatization** converts words into their dictionary base form using linguistic knowledge. For example, "running", "ran", and "runs" can all be converted to "run".

Stemming is faster but less accurate because it may produce non-dictionary words. Lemmatization is more accurate but computationally more expensive because it requires vocabulary and grammatical analysis.

```python
# Question 6:
# Write a Python program that uses regular expressions (regex) to extract all
# email addresses from the given block of text.

import re

text = """
Hello team, please contact us at support@xyz.com for technical issues,
or reach out to our HR at hr@xyz.com. You can also connect with John
at john.doe@xyz.org and jenny via jenny_clarke126@mail.co.us.
For partnership inquiries, email partners@xyz.biz.
"""

pattern = r'[a-zA-Z0-9._%+-]+@[a-zA-Z0-9.-]+\.[a-zA-Z]+'

emails = re.findall(pattern, text)

print("Extracted Email Addresses:")
for email in emails:
    print(email)
```

```python
# Question 7:
# Perform string tokenization and frequency distribution using Python and NLTK

import nltk
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

# Download required tokenizer resources
nltk.download('punkt')
nltk.download('punkt_tab')

text = """Natural Language Processing (NLP) is a fascinating field that combines linguistics,
computer science, and artificial intelligence. It enables machines to understand,
interpret, and generate human language. Applications of NLP include chatbots,
sentiment analysis, and machine translation. As technology advances, the role of NLP
in modern solutions is becoming increasingly critical."""

tokens = word_tokenize(text.lower())

print("Tokens:")
print(tokens)

freq_dist = FreqDist(tokens)

print("\nTop 10 Most Frequent Words:")
print(freq_dist.most_common(10))
```

```python
# Question 9:
# Train a simple Word2Vec model using Gensim.

!pip install gensim nltk

from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize

dataset = [
 "Natural language processing enables computers to understand human language",
 "Word embeddings are a type of word representation that allows words with similar meaning to have similar representation",
 "Word2Vec is a popular word embedding technique used in many NLP applications",
 "Text preprocessing is a critical step before training word embeddings",
 "Tokenization and normalization help clean raw text for modeling"
]

tokenized_data = [word_tokenize(sentence.lower()) for sentence in dataset]

model = Word2Vec(sentences=tokenized_data, vector_size=50, window=5, min_count=1, workers=4)

print("Vocabulary Words:")
print(list(model.wv.index_to_key))

print("\nWords similar to 'word':")
print(model.wv.most_similar("word"))
```

```python
# Question 10:
# Outline an NLP pipeline to analyze customer feedback at a fintech startup.

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import string

nltk.download('punkt')
nltk.download('stopwords')

reviews = [
    "The app is very easy to use and customer support is great",
    "Transaction failed multiple times, very frustrating experience",
    "Excellent service and fast payment processing",
    "The interface is confusing and slow"
]

stop_words = set(stopwords.words('english'))

processed_reviews = []

for review in reviews:
    tokens = word_tokenize(review.lower())

    tokens = [word for word in tokens if word not in stop_words and word not in string.punctuation]

    processed_reviews.append(tokens)

print("Processed Reviews:")
for review in processed_reviews:
    print(review)
```
