```python

from pymongo import MongoClient

# ==============================
# 1. CONNECT TO MONGODB ATLAS
# ==============================
# STEP 1: Put your real Atlas connection string here.
# Example from Atlas (may look like this):
# "mongodb+srv://<username>:<password>@cluster0.xxxxxx.mongodb.net/?retryWrites=true&w=majority&appName=Cluster0"

connection_string = "mongodb+srv://akshitasankhwar24_db_user:ZZ4GSv0hiS3m1t1N@cluster0.ehmq1pw.mongodb.net/?appName=Cluster0"

# Connect to Atlas
client = MongoClient(connection_string)

# Use your existing DB and collection
db = client["pw"]
collection = db["assignment"]

# Small helper for pretty section titles
def print_sep(title):
    print("\n" + "=" * 60)
    print(title)
    print("=" * 60)

# Test the connection (optional but useful)
try:
    client.admin.command("ping")
    print("✅ Connected to MongoDB successfully.")
except Exception as e:
    print("❌ Connection failed:", e)


# ==============================
# 2. Retrieve and print ALL documents
# ==============================
print_sep("All documents in 'assisgnment' collection")
try:
    for doc in collection.find():
        print(doc)
except Exception as e:
    print("Error while fetching documents:", e)


# ==============================
# 3. Count and display total number of documents
# ==============================
print_sep("Total number of documents in 'assisgnment'")
try:
    total_docs = collection.count_documents({})
    print("Total documents:", total_docs)
except Exception as e:
    print("Error while counting documents:", e)


# ==============================
# 4. Fetch all orders from the 'West' region
# ==============================
print_sep("All orders from Region = 'West'")
try:
    west_orders = collection.find({"Region": "West"})
    for order in west_orders:
        print(order)
except Exception as e:
    print("Error while fetching West region orders:", e)


# ==============================
# 5. Find orders where Sales is greater than 500
# ==============================
print_sep("Orders where Sales > 500")
try:
    high_sales_orders = collection.find({"Sales": {"$gt": 500}})
    for order in high_sales_orders:
        print(order)
except Exception as e:
    print("Error while fetching high sales orders:", e)


# ==============================
# 6. Fetch the top 3 orders with the highest Profit
# ==============================
print_sep("Top 3 orders with highest Profit")
try:
    top_profit_orders = collection.find().sort("Profit", -1).limit(3)
    for order in top_profit_orders:
        print(order)
except Exception as e:
    print("Error while fetching top profit orders:", e)


# ==============================
# 7. Update Ship Mode 'First Class' -> 'Premium Class'
# ==============================
print_sep("Updating Ship Mode from 'First Class' to 'Premium Class'")
try:
    update_result = collection.update_many(
        {"Ship Mode": "First Class"},
        {"$set": {"Ship Mode": "Premium Class"}}
    )
    print("Matched documents:", update_result.matched_count)
    print("Modified documents:", update_result.modified_count)
except Exception as e:
    print("Error while updating Ship Mode:", e)


# ==============================
# 8. Delete all orders where Sales < 50
# ==============================
print_sep("Deleting orders where Sales < 50")
try:
    delete_result = collection.delete_many({"Sales": {"$lt": 50}})
    print("Deleted documents:", delete_result.deleted_count)
except Exception as e:
    print("Error while deleting low sales orders:", e)


# ==============================
# 9. Aggregation: total sales per Region
# ==============================
print_sep("Total Sales per Region")
try:
    pipeline_region_sales = [
        {
            "$group": {
                "_id": "$Region",
                "totalSales": {"$sum": "$Sales"}
            }
        },
        {
            "$sort": {"totalSales": -1}
        }
    ]

    for result in collection.aggregate(pipeline_region_sales):
        print(f"Region: {result['_id']}, Total Sales: {result['totalSales']}")
except Exception as e:
    print("Error while aggregating sales per region:", e)


# ==============================
# 10. Distinct Ship Modes
# ==============================
print_sep("Distinct Ship Modes")
try:
    ship_modes = collection.distinct("Ship Mode")
    for mode in ship_modes:
        print(mode)
except Exception as e:
    print("Error while fetching distinct Ship Modes:", e)


# ==============================
# 11. Number of orders for each Category
# ==============================
print_sep("Number of orders for each Category")
try:
    pipeline_category_count = [
        {
            "$group": {
                "_id": "$Category",
                "orderCount": {"$sum": 1}
            }
        },
        {
            "$sort": {"orderCount": -1}
        }
    ]

    for result in collection.aggregate(pipeline_category_count):
        print(f"Category: {result['_id']}, Number of Orders: {result['orderCount']}")
except Exception as e:
    print("Error while counting orders per category:", e)



```
