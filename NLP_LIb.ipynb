# Question 1: Compare and contrast NLTK and spaCy in terms of features, ease of use, and performance.

**Answer:**

NLTK (Natural Language Toolkit) and spaCy are two popular Python libraries used for Natural Language Processing.

NLTK is mainly designed for research and educational purposes. It provides a wide range of linguistic tools including tokenization, stemming, parsing, tagging, and corpus analysis. It also contains many built-in datasets which help researchers experiment with language processing techniques.

spaCy is designed for production-level applications and focuses on speed and efficiency. It provides optimized NLP pipelines and pre-trained models for tasks such as named entity recognition, dependency parsing, and part-of-speech tagging.

| Feature | NLTK | spaCy |
|------|------|------|
| Primary Purpose | Research and learning | Industrial NLP applications |
| Speed | Slower | Faster |
| Ease of Use | Requires more manual setup | Easy to use with built-in pipelines |
| Pretrained Models | Limited | Extensive |
| Performance | Good for experimentation | Optimized for large-scale systems |

In summary, NLTK is ideal for learning and experimentation, while spaCy is more suitable for building real-world NLP applications.

# Question 4: Describe the architecture and functioning of a Recurrent Neural Network (RNN).

**Answer:**

A Recurrent Neural Network (RNN) is a type of neural network designed to handle sequential data such as text, speech, or time series data.

Unlike traditional neural networks, RNNs have connections that form cycles, allowing information to persist across time steps. This enables the model to remember previous inputs when processing new data.

The architecture of an RNN typically includes:

- Input layer – receives the sequence data  
- Hidden layer – maintains a hidden state that stores information from previous time steps  
- Output layer – generates predictions based on the hidden state  

At each time step, the network takes the current input and the previous hidden state to compute the new hidden state.

This ability to retain previous information allows RNNs to capture sequential relationships in text, making them useful for tasks such as language modeling, machine translation, speech recognition, and text generation.

However, traditional RNNs suffer from the vanishing gradient problem, which makes learning long-term dependencies difficult.

# Question 5: What is the key difference between LSTM and GRU networks in NLP applications?

**Answer:**

LSTM (Long Short-Term Memory) and GRU (Gated Recurrent Unit) are advanced types of Recurrent Neural Networks designed to solve the vanishing gradient problem and capture long-term dependencies in sequential data.

LSTM networks use three gates:

- Input gate  
- Forget gate  
- Output gate  

These gates control how information flows through the network and allow it to retain long-term information effectively.

GRU networks simplify this structure by using only two gates:

- Reset gate  
- Update gate  

Because GRUs have fewer gates and parameters, they are computationally faster and easier to train.

| Feature | LSTM | GRU |
|------|------|------|
| Number of Gates | 3 | 2 |
| Complexity | More complex | Simpler |
| Training Speed | Slower | Faster |
| Performance | Very powerful | Comparable to LSTM |

Both LSTM and GRU networks are widely used in NLP applications such as sentiment analysis, machine translation, and text generation.

```python
# Question 7:
# Tokenization and frequency distribution using NLTK

import nltk
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

nltk.download('punkt')
nltk.download('punkt_tab')

text = """Natural Language Processing (NLP) is a fascinating field that combines linguistics,
computer science, and artificial intelligence. It enables machines to understand,
interpret, and generate human language. Applications of NLP include chatbots,
sentiment analysis, and machine translation. As technology advances, the role of NLP
in modern solutions is becoming increasingly critical."""

tokens = word_tokenize(text.lower())

print("Tokens:")
print(tokens)

freq = FreqDist(tokens)

print("\nTop 10 Most Frequent Words:")
print(freq.most_common(10))
```

```python
# Question 8:
# Basic LSTM model for text classification

!pip install tensorflow

import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# Dataset
texts = [
"I love this project",
"This is an amazing experience",
"I hate waiting in line",
"This is the worst service",
"Absolutely fantastic!"
]

labels = [1,1,0,0,1]

# Convert labels to numpy array
labels = np.array(labels)

# Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)

sequences = tokenizer.texts_to_sequences(texts)

# Padding
padded = pad_sequences(sequences, padding='post')

# Build LSTM Model
model = Sequential()
model.add(Embedding(input_dim=100, output_dim=16, input_length=padded.shape[1]))
model.add(LSTM(16))
model.add(Dense(1, activation='sigmoid'))

# Compile model
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

# Train model
model.fit(padded, labels, epochs=10, verbose=1)

print("Model training completed.")
```

```python
# Question 9:
# spaCy NLP pipeline

!pip install spacy
!python -m spacy download en_core_web_sm

import spacy

nlp = spacy.load("en_core_web_sm")

text = """Homi Jehangir Bhaba was an Indian nuclear physicist who played a key role in the
development of India’s atomic energy program. He was the founding director of the
Tata Institute of Fundamental Research (TIFR) and was instrumental in establishing
the Atomic Energy Commission of India."""

doc = nlp(text)

print("Tokens and Lemmas:")
for token in doc:
    print(token.text,"->",token.lemma_)

print("\nNamed Entities:")
for ent in doc.ents:
    print(ent.text,"-",ent.label_)
```

```python
# Question 10:
# Simple NLP preprocessing example for chatbot input

import spacy

nlp = spacy.load("en_core_web_sm")

user_input = "I feel very anxious and stressed today."

doc = nlp(user_input)

print("Tokens:")
for token in doc:
    print(token.text)

print("\nLemmas:")
for token in doc:
    print(token.lemma_)

print("\nEntities:")
for ent in doc.ents:
    print(ent.text, ent.label_)
```
