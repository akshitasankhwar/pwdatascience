## Question 1: What is a random variable in probability theory?

**Answer:**

A random variable is a variable whose value is determined by the outcome of a random experiment. It represents numerical outcomes of a probabilistic event. Random variables allow us to quantify uncertain outcomes and analyze them using probability theory.

For example, if we roll a six-sided die, the number that appears on the die (1–6) can be considered a random variable because the outcome is uncertain.

---

## Question 2: What are the types of random variables?

**Answer:**

There are two main types of random variables:

1. **Discrete Random Variable** – It takes a finite or countable number of possible values. Examples include the number of students in a classroom or the number of heads obtained when flipping coins.

2. **Continuous Random Variable** – It can take any value within a given range and is usually measured rather than counted. Examples include height, weight, temperature, or time.

## Question 3: Explain the difference between discrete and continuous distributions.

**Answer:**

Discrete distributions represent probabilities of discrete random variables. The values are countable and usually represented by probability mass functions (PMF). Examples include the Binomial and Poisson distributions.

Continuous distributions represent probabilities of continuous random variables and are described using probability density functions (PDF). The values fall within an interval and cannot be counted individually. Examples include the Normal and Exponential distributions.

## Question 4: What is a binomial distribution, and how is it used in probability?

**Answer:**

A binomial distribution is a discrete probability distribution that models the number of successes in a fixed number of independent trials. Each trial has only two possible outcomes: success or failure, and the probability of success remains constant.

For example, if we flip a coin 10 times and count the number of heads, the result follows a binomial distribution.

## Question 5: What is the standard normal distribution, and why is it important?

**Answer:**

The standard normal distribution is a special case of the normal distribution with a mean of 0 and a standard deviation of 1. It is often used in statistical analysis because many real-world variables approximately follow a normal distribution.

It is important because it allows researchers to standardize data using z-scores and calculate probabilities using standard normal tables.

## Question 6: What is the Central Limit Theorem (CLT), and why is it critical in statistics?

**Answer:**

The Central Limit Theorem states that the sampling distribution of the sample mean will approach a normal distribution as the sample size increases, regardless of the population distribution.

This theorem is critical because it allows statisticians to make inferences about population parameters using sample data. It forms the basis for many statistical methods, including hypothesis testing and confidence interval estimation.

## Question 7: What is the significance of confidence intervals in statistical analysis?

**Answer:**

A confidence interval is a range of values used to estimate an unknown population parameter. It provides an interval within which the true parameter is likely to lie with a certain level of confidence, such as 95%.

Confidence intervals help measure the reliability and precision of statistical estimates and are widely used in research and data analysis.

## Question 8: What is the concept of expected value in a probability distribution?

**Answer:**

The expected value is the average value that a random variable is expected to take over many repeated experiments. It is calculated by multiplying each possible outcome by its probability and summing the results.

Expected value provides a measure of the central tendency of a probability distribution and helps in decision-making under uncertainty.

```python
# Question 9:
# Generate 1000 random numbers from a normal distribution with mean = 50 and std = 5
# Compute mean and standard deviation using NumPy
# Draw a histogram to visualize the distribution

import numpy as np
import matplotlib.pyplot as plt

# Generate random numbers
data = np.random.normal(loc=50, scale=5, size=1000)

# Compute statistics
mean_value = np.mean(data)
std_value = np.std(data)

print("Mean:", mean_value)
print("Standard Deviation:", std_value)

# Plot histogram
plt.hist(data, bins=30)
plt.title("Normal Distribution Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

## Question 10: Applying Central Limit Theorem to Estimate Average Sales

**Answer:**

The Central Limit Theorem states that the sampling distribution of the sample mean approaches a normal distribution as the sample size increases. This allows us to estimate the population mean using sample data.

To estimate the average sales with a 95% confidence interval:

1. Calculate the sample mean of the sales data.
2. Calculate the sample standard deviation.
3. Compute the standard error (SE) using the formula:
   SE = standard deviation / √n
4. Use the Z-score for 95% confidence level (1.96).
5. Compute the confidence interval:
   Mean ± (Z × SE)

This interval gives the range within which the true average sales are expected to fall with 95% confidence.

```python
# Question 10:
# Compute mean sales and 95% confidence interval using Central Limit Theorem

import numpy as np
import scipy.stats as stats

daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

# Convert to array
sales = np.array(daily_sales)

# Mean
mean_sales = np.mean(sales)

# Standard Error
std_error = stats.sem(sales)

# 95% Confidence Interval
confidence_interval = stats.t.interval(
    confidence=0.95,
    df=len(sales)-1,
    loc=mean_sales,
    scale=std_error
)

print("Mean Sales:", mean_sales)
print("95% Confidence Interval:", confidence_interval)
```
