# Question 1: What is the role of filters and feature maps in Convolutional Neural Network (CNN)?

**Answer:**

In a Convolutional Neural Network (CNN), filters and feature maps play a key role in extracting important patterns from input images.

A **filter (or kernel)** is a small matrix of weights that slides across the input image during the convolution operation. The purpose of the filter is to detect specific patterns in the image such as edges, textures, corners, or shapes.

When a filter is applied to the input image, it performs element-wise multiplication and summation to produce a new matrix called a **feature map**.

A **feature map** represents the output of applying a filter to the input image. It highlights where certain features are detected in the image.

Key roles:
- Filters detect patterns and features in images
- Feature maps store the detected features
- Multiple filters allow CNNs to learn many different image features

This ability to automatically learn features is one of the main strengths of CNNs.

# Question 2: Explain the concepts of padding and stride in CNNs. How do they affect the output dimensions of feature maps?

**Answer:**

Padding and stride are important parameters in the convolution operation of CNNs.

**Padding**

Padding refers to adding extra pixels (usually zeros) around the border of the input image before applying convolution. This helps preserve spatial dimensions and ensures that edge pixels are considered during convolution.

Types of padding:
- **Valid padding:** No padding is applied
- **Same padding:** Padding is added so that the output size remains the same as the input

**Stride**

Stride refers to the number of pixels the filter moves across the image during convolution.

- Stride = 1 → filter moves one pixel at a time
- Stride = 2 → filter jumps two pixels

**Effect on output size**

Increasing stride reduces the output feature map size, while padding helps maintain larger output dimensions.

Output size formula:

Output = (Input − Filter + 2 × Padding) / Stride + 1

# Question 3: Define receptive field in the context of CNNs. Why is it important for deep architectures?

**Answer:**

The **receptive field** of a neuron in a CNN refers to the region of the input image that influences the output of that neuron.

In early layers of a CNN, the receptive field is small and focuses on simple features such as edges or textures. As the network becomes deeper, the receptive field grows larger, allowing neurons to capture more complex patterns and structures.

Importance of receptive fields:

- Allows CNNs to capture local and global image features
- Helps the network understand spatial relationships
- Larger receptive fields help detect complex patterns such as objects or faces

Deep CNN architectures increase the receptive field size through multiple convolution and pooling layers, allowing them to learn hierarchical features.

# Question 4: Discuss how filter size and stride influence the number of parameters in a CNN.

**Answer:**

The number of parameters in a CNN layer depends mainly on the **filter size, number of filters, and input channels**.

The number of parameters in a convolutional layer is calculated as:

Parameters = (Filter Height × Filter Width × Input Channels) × Number of Filters + Bias

**Effect of Filter Size**

Larger filters contain more parameters, increasing the computational cost and memory usage. Smaller filters reduce the number of parameters and improve efficiency.

For example:
- 3×3 filters are commonly used because they balance performance and efficiency.

**Effect of Stride**

Stride does not change the number of parameters directly. However, larger stride reduces the spatial size of feature maps, which decreases computation in later layers.

Therefore:
- Larger filters → more parameters
- Larger stride → smaller feature maps and lower computation

# Question 5: Compare and contrast CNN architectures like LeNet, AlexNet, and VGG.

**Answer:**

LeNet, AlexNet, and VGG are important CNN architectures developed for image classification tasks.

**LeNet (1998)**

- One of the earliest CNN architectures
- Designed for handwritten digit recognition
- Uses small convolution layers and pooling layers
- Shallow architecture with fewer parameters

**AlexNet (2012)**

- Much deeper than LeNet
- Won the ImageNet competition
- Introduced ReLU activation and dropout
- Uses larger convolution filters

**VGG (2014)**

- Very deep architecture (16 or 19 layers)
- Uses small 3×3 filters consistently
- Achieves high performance on image classification tasks
- Requires high computational resources

| Architecture | Depth | Filter Size | Performance |
|---------------|------|-------------|-------------|
| LeNet | Shallow | 5×5 | Basic tasks |
| AlexNet | Medium | 11×11, 5×5 | High performance |
| VGG | Deep | 3×3 | Very high accuracy |

VGG improved performance by increasing depth while using smaller filters.

```python
# Question 6
# CNN on MNIST using Keras

import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

(X_train,y_train),(X_test,y_test)=mnist.load_data()

X_train=X_train.reshape(-1,28,28,1)/255
X_test=X_test.reshape(-1,28,28,1)/255

model=Sequential([
Conv2D(32,(3,3),activation='relu',input_shape=(28,28,1)),
MaxPooling2D((2,2)),
Conv2D(64,(3,3),activation='relu'),
MaxPooling2D((2,2)),
Flatten(),
Dense(128,activation='relu'),
Dense(10,activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(X_train,y_train,epochs=5)

loss,acc=model.evaluate(X_test,y_test)

print("Test Accuracy:",acc)
```

```python
# Question 8
# PyTorch CNN for MNIST

import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms

transform=transforms.ToTensor()

trainset=torchvision.datasets.MNIST(root='./data',train=True,download=True,transform=transform)
trainloader=torch.utils.data.DataLoader(trainset,batch_size=64,shuffle=True)

class CNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1=nn.Conv2d(1,32,3)
        self.pool=nn.MaxPool2d(2,2)
        self.fc1=nn.Linear(32*13*13,10)

    def forward(self,x):
        x=self.pool(torch.relu(self.conv1(x)))
        x=x.view(x.size(0),-1)
        x=self.fc1(x)
        return x

model=CNN()

criterion=nn.CrossEntropyLoss()
optimizer=optim.Adam(model.parameters())

for epoch in range(2):
    for images,labels in trainloader:
        optimizer.zero_grad()
        outputs=model(images)
        loss=criterion(outputs,labels)
        loss.backward()
        optimizer.step()

print("Training completed")
```

```python
# Question 9
# Training CNN using ImageDataGenerator

import os
import numpy as np
from PIL import Image
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

# ------------------------------------------------
# STEP 1: Create dummy dataset (so code runs)
# ------------------------------------------------

os.makedirs("dataset/class1", exist_ok=True)
os.makedirs("dataset/class2", exist_ok=True)

for i in range(10):

    img = np.random.randint(0,255,(224,224,3),dtype=np.uint8)
    Image.fromarray(img).save(f"dataset/class1/img{i}.jpg")

    img = np.random.randint(0,255,(224,224,3),dtype=np.uint8)
    Image.fromarray(img).save(f"dataset/class2/img{i}.jpg")

print("Dummy dataset created")

# ------------------------------------------------
# STEP 2: ImageDataGenerator
# ------------------------------------------------

datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

train_data = datagen.flow_from_directory(
    "dataset/",
    target_size=(224,224),
    batch_size=4,
    class_mode='binary',
    subset='training'
)

val_data = datagen.flow_from_directory(
    "dataset/",
    target_size=(224,224),
    batch_size=4,
    class_mode='binary',
    subset='validation'
)

# ------------------------------------------------
# STEP 3: CNN Model
# ------------------------------------------------

model = Sequential([
    Conv2D(32,(3,3),activation='relu',input_shape=(224,224,3)),
    MaxPooling2D((2,2)),
    Flatten(),
    Dense(128,activation='relu'),
    Dense(1,activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# ------------------------------------------------
# STEP 4: Train Model
# ------------------------------------------------

history = model.fit(
    train_data,
    epochs=3,
    validation_data=val_data
)

print("Training completed")
```
