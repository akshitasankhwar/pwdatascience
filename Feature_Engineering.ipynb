1.  ### What is a parameter?
    A **parameter** is a configuration variable that's **internal** to a machine learning model. Its value is learned automatically from the data during the training process. For example, in a linear regression model, y = mx + c, the slope (m) and intercept (c) are parameters.

-----

2.  ### What is correlation?
    **Correlation** is a statistical measure that shows how strongly two variables are linearly related. It's measured on a scale from **-1 to +1**. A value of +1 means a perfect positive relationship, -1 means a perfect negative relationship, and 0 means no linear relationship.

-----

3.  ### What does negative correlation mean?
    **Negative correlation** means there's an **inverse relationship** between two variables. When one variable increases, the other tends to decrease. For example, as the hours spent playing video games increase, exam scores might decrease.

-----

4.  ### Define Machine Learning. What are the main components in Machine Learning?

    **Machine Learning (ML)** is a field of AI that gives computers the ability to learn from data and make predictions or decisions without being explicitly programmed.

    The main components are:

      * **Data**: The information used by the model to learn.
      * **Model**: The algorithm (e.g., linear regression, neural network) that learns patterns.
      * **Loss Function**: A method to measure the model's error or how "wrong" its predictions are.
      * **Optimizer**: An algorithm that adjusts the model's internal parameters to minimize the loss.

-----

5.  ### How does loss value help in determining whether the model is good or not?
    The **loss value** quantifies the model's error. A **lower loss value generally indicates a better model** because its predictions are closer to the actual, true values. During training, the goal is to adjust the model's parameters to find the minimum possible loss.

-----

6.  ### What are continuous and categorical variables?
      * **Continuous Variables**: These are numerical variables that can take any value within a given range. They can be measured. Examples include **height**, **temperature**, and **price**.
      * **Categorical Variables**: These represent distinct categories or labels. Examples include **gender** ("Male", "Female"), **color** ("Red", "Green"), or **city**.

-----

7.  ### How do we handle categorical variables in Machine Learning? What are the common techniques?

    We must convert categorical variables into numbers, a process called **encoding**.

    Common techniques are:

      * **One-Hot Encoding**: Creates a new binary (0 or 1) column for each category. It's best for nominal data (where order doesn't matter).
      * **Label Encoding**: Assigns a unique integer to each category (e.g., "Red" -\> 0, "Green" -\> 1). It's suitable for ordinal data (where order matters) or for tree-based models.

-----

8.  ### What do you mean by training and testing a dataset?
      * **Training Set**: This is the portion of the dataset (usually 70-80%) used to **train the model**. The model learns patterns and relationships from this data.
      * **Testing Set**: This is the unseen portion of the dataset used to **evaluate the model's performance**. It gives an unbiased estimate of how the model will perform on new, real-world data.

-----

    `sklearn.preprocessing` is a module in Python's Scikit-learn library that provides functions and classes for common data preprocessing tasks. This includes tools for **feature scaling** (`StandardScaler`, `MinMaxScaler`) and **encoding categorical variables** (`OneHotEncoder`).

10. ### What is a Test set?
    A **Test Set** is a subset of the dataset that is held back and not used during model training. Its primary purpose is to provide an **unbiased evaluation** of the final model's performance on unseen data.

12. ### How do you approach a Machine Learning problem?
    A typical approach involves these steps:
    1.  **Define the Problem**: Understand the goal and success metrics.
    2.  **Gather Data**: Collect the necessary data.
    3.  **Exploratory Data Analysis (EDA)**: Analyze and visualize data to find patterns and issues.
    4.  **Data Preprocessing**: Clean the data, handle missing values, encode, and scale features.
    5.  **Model Selection**: Choose a suitable algorithm.
    6.  **Model Training**: Fit the model to the training data.
    7.  **Model Evaluation**: Test the model's performance on the test set.
    8.  **Tuning & Optimization**: Improve the model by adjusting its settings.
    9.  **Deployment**: Make the model available for use.

17. ### What is causation? Explain the difference between correlation and causation with an example.

    **Causation** means that one event is the direct result of another (a cause-and-effect relationship). The key difference is that **correlation does not imply causation**.

      * **Example**: Ice cream sales and drowning incidents are positively correlated.
      * **Correlation**: When ice cream sales go up, drowning incidents also go up.
      * **Does one cause the other?**: No. Eating ice cream doesn't cause drowning.
      * **The real cause (lurking variable)**: **Hot weather**. Hot weather causes more people to buy ice cream AND causes more people to go swimming, which increases the risk of drowning.

18. ### What is an Optimizer? What are different types of optimizers?

    An **optimizer** is an algorithm used to change the model's internal parameters (like weights) to minimize the loss function.

    Different types include:

      * **Gradient Descent (GD)**: Computes the gradient using the **entire training dataset** for each update. It's accurate but slow on large datasets.
      * **Stochastic Gradient Descent (SGD)**: Updates the parameters after seeing **each single training example**. It's much faster but the updates are noisy.
      * **Adam (Adaptive Moment Estimation)**: An efficient, popular optimizer that adapts the learning rate for each parameter. It's often the default choice for deep learning.

23. ### What is feature scaling? How does it help in Machine Learning?

    **Feature scaling** is a preprocessing technique that standardizes the range of features in the data.

    It helps because:

      * **Improves Convergence**: Algorithms that use gradient descent (like linear regression and neural networks) train much faster.
      * **Equalizes Feature Importance**: It prevents features with larger numeric ranges from dominating those with smaller ranges, which is crucial for distance-based algorithms like KNN.

-----

24. ### How do we perform scaling in Python?

    We use classes from `sklearn.preprocessing`, like `StandardScaler` or `MinMaxScaler`.

    ```python
    from sklearn.preprocessing import StandardScaler

    # Create a scaler object
    scaler = StandardScaler()

    # Fit on training data and transform it
    X_train_scaled = scaler.fit_transform(X_train)

    # Use the SAME scaler to transform the test data (to avoid data leakage)
    X_test_scaled = scaler.transform(X_test)
    ```

27. ### Explain data encoding.
    **Data encoding** is the process of converting data from one format to another. In machine learning, it specifically refers to converting **categorical variables** (like text labels) into a **numerical format**. This is necessary because most machine learning algorithms can only process numbers, not text. Common techniques include One-Hot Encoding and Label Encoding.
