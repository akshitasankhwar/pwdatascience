## Question 1: What is hypothesis testing in statistics?

**Answer:**

Hypothesis testing is a statistical method used to make decisions or draw conclusions about a population based on sample data. It involves formulating two competing hypotheses and using statistical tests to determine whether there is enough evidence to reject the null hypothesis.

The process generally includes defining the null hypothesis, selecting a significance level, calculating a test statistic, and comparing the result with a critical value or p-value. Hypothesis testing helps researchers determine whether observed results are statistically significant or occurred by chance.

---

## Question 2: What is the null hypothesis, and how does it differ from the alternative hypothesis?

**Answer:**

The null hypothesis (H₀) is the default assumption that there is no significant difference or relationship between variables in a population. It represents the idea that any observed effect in the data is due to random chance.

The alternative hypothesis (H₁ or Ha) is the statement that contradicts the null hypothesis. It suggests that there is a significant difference or relationship between variables.

For example, if we test whether a new medicine improves recovery time:
- Null Hypothesis: The medicine has no effect.
- Alternative Hypothesis: The medicine improves recovery time.

---

## Question 3: Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.

**Answer:**

The significance level (α) is the probability of rejecting the null hypothesis when it is actually true. It represents the threshold used to determine whether the result of a statistical test is statistically significant.

Common significance levels include 0.05 (5%) and 0.01 (1%). If the p-value obtained from a statistical test is less than the significance level, the null hypothesis is rejected. Otherwise, the null hypothesis is not rejected.

Thus, the significance level helps determine whether the observed results are strong enough to conclude that a real effect exists.

---

## Question 4: What are Type I and Type II errors? Give examples of each.

**Answer:**

Type I Error occurs when the null hypothesis is rejected even though it is actually true. This is also known as a false positive.

Example: Concluding that a medicine works when it actually does not.

Type II Error occurs when the null hypothesis is not rejected even though it is false. This is also known as a false negative.

Example: Concluding that a medicine does not work when it actually does.

Both errors are important considerations in statistical hypothesis testing.

## Question 5: What is the difference between a Z-test and a T-test? Explain when to use each.

**Answer:**

A Z-test is used when the population standard deviation is known and the sample size is large (typically greater than 30). It assumes that the sampling distribution follows a normal distribution.

A T-test is used when the population standard deviation is unknown and the sample size is small. The T-distribution is used instead of the normal distribution to account for the additional uncertainty.

In practice, T-tests are more commonly used because the population standard deviation is rarely known.

```python
# Question 7:
# Implement hypothesis testing using Z-statistics

import numpy as np
from scipy import stats

sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1)
n = len(sample_data)

# Hypothesized population mean
mu = 50

z_score = (sample_mean - mu) / (sample_std / np.sqrt(n))
p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))

print("Sample Mean:", sample_mean)
print("Z-score:", z_score)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
# Question 8:
# Simulate normal distribution data and calculate 95% confidence interval

import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

data = np.random.normal(loc=100, scale=15, size=500)

mean = np.mean(data)
std_error = stats.sem(data)

confidence_interval = stats.t.interval(
    confidence=0.95,
    df=len(data)-1,
    loc=mean,
    scale=std_error
)

print("Mean:", mean)
print("95% Confidence Interval:", confidence_interval)

plt.hist(data, bins=30)
plt.title("Normal Distribution Simulation")
plt.show()
```

```python
# Question 9:
# Calculate Z-scores and visualize standardized data

import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 100)

mean = np.mean(data)
std = np.std(data)

z_scores = (data - mean) / std

print("First 10 Z-scores:", z_scores[:10])

plt.hist(z_scores, bins=20)
plt.title("Histogram of Z-Scores")
plt.xlabel("Z-score")
plt.ylabel("Frequency")
plt.show()
```

### Explanation

Z-scores represent how many standard deviations a data point is from the mean of the dataset.

- A Z-score of 0 means the value is exactly equal to the mean.
- A positive Z-score indicates the value is above the mean.
- A negative Z-score indicates the value is below the mean.

Standardizing data using Z-scores helps compare values from different datasets and identify outliers.
