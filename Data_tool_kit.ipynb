**Q1: What is NumPy, and why is it widely used in Python?**  
NumPy is a Python library for numerical computing. It is widely used because it provides fast, efficient, and easy-to-use arrays and mathematical operations, making it ideal for scientific computing and data analysis.

**Q2: How does broadcasting work in NumPy?**  
Broadcasting allows NumPy to perform arithmetic operations on arrays of different shapes by automatically expanding the smaller array to match the shape of the larger one.

**Q3: What is a Pandas DataFrame?**  
A Pandas DataFrame is a 2D labeled data structure (like a table) with rows and columns, used for storing and manipulating structured data.

**Q4: Explain the use of the groupby() method in Pandas.**  
`groupby()` is used to split data into groups based on a column, then apply functions like sum, mean, or count to each group for analysis.

**Q5: Why is Seaborn preferred for statistical visualizations?**  
Seaborn is preferred because it provides high-level functions, built-in themes, and integration with Pandas, making it easier to create attractive and informative statistical plots.

**Q6: What are the differences between NumPy arrays and Python lists?**  
- NumPy arrays are faster and more memory-efficient.  
- They support vectorized operations, unlike lists.  
- Lists can store mixed data types, arrays are homogeneous.

**Q7: What is a heatmap, and when should it be used?**  
A heatmap is a graphical representation of data where values are shown as colors. It is used to visualize correlations, distributions, or intensity of data.

**Q8: What does the term “vectorized operation” mean in NumPy?**  
Vectorized operations allow applying mathematical operations directly on entire arrays without explicit loops, improving performance and readability.

**Q9: How does Matplotlib differ from Plotly?**  
- Matplotlib produces static plots, while Plotly creates interactive plots.  
- Plotly allows zooming/hovering, while Matplotlib is better for publication-quality static visuals.

**Q10: What is the significance of hierarchical indexing in Pandas?**  
Hierarchical indexing allows multiple levels of indexing on rows or columns, making it easier to handle higher-dimensional data in a 2D DataFrame.

**Q11: What is the role of Seaborn’s pairplot() function?**  
`pairplot()` plots pairwise relationships in a dataset, showing scatterplots for variable combinations and distributions for individual variables.

**Q12: What is the purpose of the describe() function in Pandas?**  
`describe()` provides summary statistics (mean, median, min, max, std, etc.) for numeric columns in a DataFrame.

**Q13: Why is handling missing data important in Pandas?**  
Missing data can lead to errors, biased results, or incorrect analysis, so it must be handled by filling, dropping, or imputing values.

**Q14: What are the benefits of using Plotly for data visualization?**  
- Interactive plots with zoom/hover options.  
- Easy integration with Pandas and Jupyter/Colab.  
- Wide variety of visualization types.

**Q15: How does NumPy handle multidimensional arrays?**  
NumPy uses the `ndarray` object to handle multidimensional arrays, where each dimension is called an axis, allowing efficient operations on n-dimensional data.

**Q16: What is the role of Bokeh in data visualization?**  
Bokeh is a Python library for creating interactive and web-ready visualizations that can be embedded in web applications.

**Q17: Explain the difference between apply() and map() in Pandas.**  
- `map()` works on Series and applies a function element-wise.  
- `apply()` works on both Series and DataFrame, applying functions row-wise or column-wise.

**Q18: What are some advanced features of NumPy?**  
- Broadcasting for arithmetic operations.  
- Linear algebra, Fourier transforms, random number generation.  
- Efficient handling of large multidimensional arrays.

**Q19: How does Pandas simplify time series analysis?**  
Pandas provides built-in support for date/time indexing, resampling, shifting, and frequency conversion, making time series analysis easier.

**Q20: What is the role of a pivot table in Pandas?**  
A pivot table summarizes data by reorganizing it into a new table format, allowing quick aggregation and comparison across categories.

**Q21: Why is NumPy’s array slicing faster than Python’s list slicing?**  
NumPy arrays store data in contiguous memory blocks and use optimized C operations, making slicing much faster compared to Python lists.

**Q22: What are some common use cases for Seaborn?**  
- Visualizing distributions (histogram, KDE).  
- Correlation heatmaps.  
- Pairwise plots for datasets.  
- Categorical plots (bar, box, violin).

```python
#Write a Pandas script to find the mean of a specific column in a DataFrame
import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [24, 30, 22, 28],
    'Salary': [50000, 60000, 55000, 65000]
}
df = pd.DataFrame(data)
mean_age = df['Age'].mean()
print("Mean Age:", mean_age)
mean_salary = df['Salary'].mean()
print("Mean Salary:", mean_salary)
```

```python
#Create a scatter plot using Matplotlib
import matplotlib.pyplot as plt
x = [5, 7, 8, 7, 6, 9, 5, 6, 7, 8]
y = [99, 86, 87, 88, 100, 86, 103, 87, 94, 78]
plt.scatter(x, y, color="blue", marker="o", s=100, alpha=0.7)
plt.xlabel("X-axis Values")
plt.ylabel("Y-axis Values")
plt.title("Simple Scatter Plot")
plt.show()
```

```python
#How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data = {
    'Math': [85, 90, 78, 92, 88],
    'Physics': [80, 85, 75, 89, 90],
    'Chemistry': [78, 82, 72, 88, 85],
    'Biology': [90, 88, 80, 85, 87]
}
df = pd.DataFrame(data)
corr_matrix = df.corr()
print("Correlation Matrix:")
print(corr_matrix)
plt.figure(figsize=(6,4))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Matrix Heatmap")
plt.show()
```

```python
#Generate a bar plot using Plotly
import plotly.express as px
import pandas as pd
data = {
    "Fruits": ["Apples", "Bananas", "Oranges", "Grapes"],
    "Sales": [50, 70, 40, 90]
}
df = pd.DataFrame(data)
fig = px.bar(df, x="Fruits", y="Sales", title="Fruit Sales", color="Fruits")
fig.show()
```

```python
#Create a DataFrame and add a new column based on an existing column
import pandas as pd
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [24, 30, 22, 28]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
df["Age_Group"] = df["Age"].apply(lambda x: "Young" if x < 25 else "Adult")
print("\nDataFrame with New Column:")
print(df)
```

```python
#Write a program to perform element-wise multiplication of two NumPy array
import numpy as np
arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([5, 6, 7, 8])
print("Array 1:", arr1)
print("Array 2:", arr2)
result = arr1 * arr2
print("Element-wise Multiplication Result:", result)
```

```python
#Create a line plot with multiple lines using Matplotlib
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 3, 5, 7, 9]
y3 = [2, 3, 4, 5, 6]
plt.plot(x, y1, label="Line 1 - y=2x", marker='o')
plt.plot(x, y2, label="Line 2 - y=2x-1", marker='s')
plt.plot(x, y3, label="Line 3 - y=x+1", marker='^')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Line Plot with Multiple Lines")
plt.legend()
plt.show()
```

```python
#Generate a Pandas DataFrame and filter rows where a column value is greater than a threshold
import pandas as pd
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [24, 30, 22, 28, 35],
    "Salary": [50000, 60000, 45000, 70000, 80000]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
filtered_df = df[df["Salary"] > 60000]
print("\nFiltered DataFrame (Salary > 60000):")
print(filtered_df)
```
