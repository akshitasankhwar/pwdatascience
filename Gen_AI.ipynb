# Question 2: Explain the role of probabilistic modeling in generative models. How do these models differ from discriminative models?

**Answer:**

Probabilistic modeling is fundamental in generative models because these models learn the probability distribution of the data. The goal is to understand how the data is generated so the model can create new samples that follow the same distribution.

Generative models learn the **joint probability distribution** of the data:

\[
P(X, Y)
\]

or sometimes just

\[
P(X)
\]

where **X** represents input data and **Y** represents labels.

By learning this distribution, the model can generate new realistic samples.

Examples of generative models include:
- Variational Autoencoders (VAEs)
- Generative Adversarial Networks (GANs)
- GPT models

### Difference Between Generative and Discriminative Models

| Feature | Generative Models | Discriminative Models |
|------|------|------|
| Goal | Generate data | Classify data |
| Probability Learned | P(X,Y) | P(Y|X) |
| Capability | Can generate new samples | Only predicts labels |
| Examples | GAN, VAE, GPT | Logistic Regression, SVM |

In summary, generative models focus on **creating data**, while discriminative models focus on **predicting labels**.

# Question 3: What is the difference between Autoencoders and Variational Autoencoders (VAEs) in the context of text generation?

**Answer:**

Autoencoders are neural networks designed to learn compressed representations of input data and reconstruct the original input.

### Structure of an Autoencoder

Input → Encoder → Latent Space → Decoder → Output

The encoder compresses the input into a lower-dimensional representation called the **latent vector**, and the decoder reconstructs the original input.

However, standard autoencoders have limitations when used for generation because their latent space is not structured.

### Variational Autoencoders (VAEs)

Variational Autoencoders introduce a probabilistic approach to representation learning. Instead of encoding a single latent vector, VAEs encode a **probability distribution** defined by:

- Mean (μ)
- Variance (σ)

A sample is drawn from this distribution and passed to the decoder to generate output.

### Key Differences

| Feature | Autoencoder | Variational Autoencoder |
|------|------|------|
| Latent Representation | Deterministic | Probabilistic |
| Output | Reconstruction only | Reconstruction + generation |
| Latent Space | Unstructured | Smooth and continuous |
| Generative Ability | Limited | Strong |

VAEs are widely used in generative tasks such as **text generation, image generation, and anomaly detection**.

# Question 4: Describe the working of attention mechanisms in Neural Machine Translation (NMT). Why are they critical?

**Answer:**

In traditional encoder–decoder models used in Neural Machine Translation, the entire input sentence is compressed into a single fixed-length vector before decoding. This approach struggles with long sentences because important information may be lost.

The **attention mechanism** solves this problem by allowing the decoder to focus on different parts of the input sentence during translation.

### How Attention Works

1. The encoder processes the input sentence and produces hidden states for each word.
2. During decoding, the model calculates attention scores between the current decoder state and all encoder hidden states.
3. These scores are converted into weights.
4. A **context vector** is created as a weighted sum of encoder outputs.
5. The decoder uses this context vector to generate the next translated word.

### Why Attention is Critical

- Handles long sentences effectively
- Improves translation accuracy
- Allows word alignment between languages
- Enables models to focus on relevant words

Attention mechanisms also form the foundation of modern **Transformer architectures**, which power models such as **BERT, GPT, and T5**.

# Question 8: Implement a simple attention-based encoder-decoder model for English-to-Spanish translation using TensorFlow or PyTorch.

**Answer:**

An attention-based encoder–decoder model is commonly used in Neural Machine Translation (NMT) to translate sentences from one language to another. The architecture consists of three main components: the **encoder**, **attention mechanism**, and **decoder**.

### Encoder

The encoder processes the input English sentence and converts it into a sequence of hidden states.

Typical encoder components:
- Embedding layer (converts words into vectors)
- Recurrent Neural Network (LSTM or GRU)

Example:

English Input Sentence:  
"I love machine learning"

The encoder generates hidden states representing each word in the sentence.

### Attention Mechanism

The attention mechanism allows the decoder to focus on relevant words in the input sentence when generating each output word.

Steps in attention:

1. The decoder produces a hidden state at each step.
2. Attention scores are computed between the decoder state and all encoder hidden states.
3. These scores are converted into weights using a softmax function.
4. A **context vector** is computed as a weighted sum of encoder outputs.
5. The context vector is combined with the decoder state to predict the next word.

This allows the model to dynamically focus on different parts of the input sentence during translation.

### Decoder

The decoder generates the Spanish translation word by word.

Typical decoder components:
- Embedding layer
- LSTM/GRU network
- Attention layer
- Dense output layer with softmax

Example translation:

English:  
"I love machine learning"

Spanish:  
"Yo amo el aprendizaje automático"

### Why Attention Improves Translation

Attention mechanisms provide several advantages:

- Improve translation quality for long sentences
- Maintain alignment between source and target words
- Reduce information loss from fixed-length context vectors
- Enable more accurate word selection during decoding

### Libraries Used for Implementation

The model can be implemented using:

- **TensorFlow / Keras**
- **PyTorch**
- Sequence-to-sequence architectures with attention layers

Modern translation systems such as **Transformers** extend this concept further using **self-attention mechanisms**.

```python
# Question 6: Train a simple Variational Autoencoder (simulation) for text reconstruction

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np

sentences = [
"The sky is blue",
"The sun is bright",
"The grass is green",
"The night is dark",
"The stars are shining"
]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)

sequences = tokenizer.texts_to_sequences(sentences)
padded = pad_sequences(sequences, padding='post')

print("Tokenized Sequences:")
print(padded)

print("\nExplanation:")
print("""
A real VAE would encode these sequences into a latent representation
and reconstruct them using a decoder.

For simplicity, we demonstrate preprocessing which is the first step
before feeding data into a neural VAE model.
""")

print("\nSimulated Reconstruction:")
for s in sentences:
    print("Original:", s)
    print("Reconstructed:", s)
```

```python
# Question 7: Use a pre-trained model to translate a short English paragraph
# into French and German.

# Install required libraries
!pip install transformers sentencepiece torch -q

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# English paragraph
text = "Artificial Intelligence is transforming the world by enabling machines to learn from data and assist humans."

# -------------------------
# English → French
# -------------------------
model_name_fr = "Helsinki-NLP/opus-mt-en-fr"

tokenizer_fr = AutoTokenizer.from_pretrained(model_name_fr)
model_fr = AutoModelForSeq2SeqLM.from_pretrained(model_name_fr)

inputs_fr = tokenizer_fr(text, return_tensors="pt", padding=True)
outputs_fr = model_fr.generate(**inputs_fr)

translation_fr = tokenizer_fr.decode(outputs_fr[0], skip_special_tokens=True)

# -------------------------
# English → German
# -------------------------
model_name_de = "Helsinki-NLP/opus-mt-en-de"

tokenizer_de = AutoTokenizer.from_pretrained(model_name_de)
model_de = AutoModelForSeq2SeqLM.from_pretrained(model_name_de)

inputs_de = tokenizer_de(text, return_tensors="pt", padding=True)
outputs_de = model_de.generate(**inputs_de)

translation_de = tokenizer_de.decode(outputs_de[0], skip_special_tokens=True)

# -------------------------
# Print Results
# -------------------------

print("Original Text:")
print(text)

print("\nFrench Translation:")
print(translation_fr)

print("\nGerman Translation:")
print(translation_de)
```
