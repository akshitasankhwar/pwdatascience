# Question 1: What are Large Language Models (LLMs) and how do they function?

**Answer:**

Large Language Models (LLMs) are advanced artificial intelligence models designed to understand, process, and generate human-like text. They are trained on extremely large datasets consisting of books, articles, websites, and other text sources.

LLMs are typically built using **Transformer architectures**, which use mechanisms such as **self-attention** to understand relationships between words in a sentence.

### How LLMs Function

1. **Pretraining**
   - The model is trained on large amounts of text data.
   - It learns statistical patterns of language by predicting the next word in a sentence.

2. **Tokenization**
   - Input text is converted into smaller units called tokens.

3. **Transformer Architecture**
   - Uses attention mechanisms to capture relationships between words.

4. **Inference**
   - When a user provides a prompt, the model predicts the most probable sequence of words as output.

Examples of LLMs include:
- GPT models
- BERT
- LLaMA
- PaLM

LLMs are widely used for tasks such as question answering, summarization, translation, and conversational AI.

# Question 5: Compare and contrast Langchain and LamaIndex. What unique problems does each solve?

**Answer:**

LangChain and LlamaIndex are frameworks designed to help developers build applications using Large Language Models.

### LangChain

LangChain is a framework that helps developers build applications powered by LLMs by connecting models with external tools, APIs, and workflows.

Key features:
- Prompt templates
- Chains and agents
- Memory for conversations
- Integration with APIs and databases

LangChain is mainly used for building **LLM-powered workflows and applications**.

### LlamaIndex

LlamaIndex focuses on connecting LLMs with **external data sources** such as documents, PDFs, databases, and knowledge bases.

Key features:
- Document indexing
- Retrieval-based question answering
- Efficient information retrieval

LlamaIndex is primarily used for **retrieval-augmented generation (RAG)**.

### Comparison

| Feature | LangChain | LlamaIndex |
|-------|-------|-------|
| Purpose | Build LLM applications | Connect LLMs with data |
| Focus | Workflows and chains | Document indexing |
| Use Case | Chatbots, agents | Document search systems |

### Summary

LangChain helps manage **LLM workflows and logic**, while LlamaIndex focuses on **efficient data retrieval from large document collections**.

```python
# Question 9: Combining LangChain with LlamaIndex for a Document-Based Q&A Chatbot

# ── Install dependencies ──────────────────────────────────────────────────────
!pip install llama-index llama-index-embeddings-huggingface sentence-transformers -q

# ── Step 1: Create a sample ML document ──────────────────────────────────────
text = """
Machine Learning is a branch of Artificial Intelligence that focuses on
building systems that learn from data.

Supervised learning uses labeled datasets to train models.
Unsupervised learning identifies patterns in unlabeled data.

Deep learning is a subset of machine learning that uses neural networks
with multiple layers to analyze complex data.

Natural Language Processing (NLP) is a field of AI that helps computers
understand, interpret, and generate human language.

Reinforcement learning is a type of machine learning where an agent learns
by interacting with an environment and receiving rewards or penalties.
"""

with open("ml_document.txt", "w") as f:
    f.write(text)

print("✅ Document created: ml_document.txt")

# ── Step 2: Imports ───────────────────────────────────────────────────────────
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader, Settings
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

# ── Step 3: Use FREE local embedding model (no API key needed) ────────────────
print("\n⏳ Loading embedding model...")
Settings.embed_model = HuggingFaceEmbedding(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
Settings.llm = None   # Disable LLM — return retrieved text directly (no OpenAI needed)
print("✅ Embedding model loaded.")

# ── Step 4: Load the document ─────────────────────────────────────────────────
documents = SimpleDirectoryReader(input_files=["ml_document.txt"]).load_data()
print(f"✅ Document loaded. Total chunks: {len(documents)}")

# ── Step 5: Build the vector index ───────────────────────────────────────────
print("\n⏳ Building vector index...")
index = VectorStoreIndex.from_documents(documents)
print("✅ Index built successfully.")

# ── Step 6: Create the query engine ──────────────────────────────────────────
query_engine = index.as_query_engine(similarity_top_k=2)

# ── Step 7: Run the chatbot ───────────────────────────────────────────────────
questions = [
    "What is machine learning?",
    "What is supervised learning?",
    "What is deep learning?",
    "What is NLP?",
    "What is reinforcement learning?"
]

print("\n" + "=" * 65)
print("         🤖 Document-Based Q&A Chatbot (LlamaIndex)")
print("=" * 65)

for q in questions:
    response = query_engine.query(q)
    print(f"\n👤 User    : {q}")
    print(f"🤖 Chatbot : {str(response).strip()}")
    print("-" * 65)

print("\n✅ Chatbot demo complete!")
```

```python
# Question 10: AI Solution for Legal Document Summarization and Retrieval
# Using LangChain + LlamaIndex for a Legal Firm

# ── Install dependencies ──────────────────────────────────────────────────────
!pip install llama-index llama-index-embeddings-huggingface sentence-transformers langchain-text-splitters -q

# ── Step 1: Create sample legal documents ────────────────────────────────────
import os
os.makedirs("legal_docs", exist_ok=True)

doc1 = """
CONTRACT AGREEMENT - Case No. 2024-001

This agreement is entered into on January 15, 2024, between Alpha Corporation
(hereinafter "Client") and Beta Legal Services (hereinafter "Provider").

TERMS AND CONDITIONS:
1. The Provider agrees to deliver legal consultation services for a period of
   12 months starting from the date of signing.
2. The Client agrees to pay a monthly retainer fee of $5,000.
3. Either party may terminate this agreement with 30 days written notice.
4. All disputes arising from this agreement shall be resolved through arbitration
   under the laws of New York State.
5. Confidentiality: Both parties agree to keep all shared information strictly
   confidential for a period of 5 years after termination.

PENALTY CLAUSE:
Breach of confidentiality will result in a penalty of $50,000 and immediate
termination of the agreement.
"""

doc2 = """
COURT JUDGMENT - Case No. 2023-CR-456

Date: March 10, 2023
Court: District Court of California
Judge: Hon. Patricia Williams

PARTIES INVOLVED:
- Plaintiff: Green Energy Ltd.
- Defendant: Carbon Industries Inc.

SUMMARY OF CASE:
Green Energy Ltd. filed a lawsuit against Carbon Industries Inc. for violation
of environmental regulations under the Clean Air Act of 1990. The plaintiff
alleged that the defendant discharged harmful pollutants exceeding the legally
permitted limit for 18 consecutive months.

JUDGMENT:
The court finds the defendant GUILTY of environmental violations.

PENALTIES IMPOSED:
1. A fine of $2,000,000 to be paid within 90 days.
2. Mandatory installation of emission control systems within 6 months.
3. Monthly environmental audits for the next 3 years.
4. Compensation of $500,000 to affected local communities.

The defendant has the right to appeal this judgment within 30 days.
"""

doc3 = """
LEGAL POLICY DOCUMENT - Employment Law Guidelines 2024

SECTION 1: WRONGFUL TERMINATION
An employee may file a wrongful termination claim if dismissed without just cause,
in violation of an employment contract, or due to discrimination based on race,
gender, age, religion, or disability.

SECTION 2: WORKPLACE HARASSMENT POLICY
All organizations with more than 10 employees must maintain a formal anti-harassment
policy. Employees must report harassment to HR within 60 days of occurrence.
Employers failing to act on harassment complaints may face fines up to $100,000.

SECTION 3: COMPENSATION AND BENEFITS
- Minimum wage compliance is mandatory under federal and state laws.
- Overtime pay at 1.5x the regular rate is required for hours worked beyond 40 per week.
- Employees are entitled to 12 weeks of unpaid family leave under FMLA.

SECTION 4: DATA PRIVACY IN EMPLOYMENT
Employers must comply with GDPR and CCPA when handling employee personal data.
Unauthorized access or sharing of employee data may result in penalties up to $250,000.
"""

with open("legal_docs/contract_agreement.txt", "w") as f:
    f.write(doc1)
with open("legal_docs/court_judgment.txt", "w") as f:
    f.write(doc2)
with open("legal_docs/employment_law.txt", "w") as f:
    f.write(doc3)

print("✅ 3 Legal documents created inside 'legal_docs/' folder.")

# ── Step 2: Imports ───────────────────────────────────────────────────────────
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader, Settings
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

# ✅ FIXED: New correct import path for LangChain text splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter

# ── Step 3: Load and preprocess documents using LangChain text splitter ───────
print("\n⏳ Loading and preprocessing legal documents with LangChain...")

raw_texts = []
file_names = []

for filename in os.listdir("legal_docs"):
    if filename.endswith(".txt"):
        with open(f"legal_docs/{filename}", "r") as f:
            raw_texts.append(f.read())
            file_names.append(filename)

# LangChain: Split large documents into manageable chunks
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
    separators=["\n\n", "\n", ".", " "]
)

all_chunks = []
for i, text in enumerate(raw_texts):
    chunks = splitter.split_text(text)
    all_chunks.extend(chunks)
    print(f"   📄 {file_names[i]} → split into {len(chunks)} chunks")

print(f"\n✅ Total chunks after LangChain splitting: {len(all_chunks)}")

# ── Step 4: Document Summarization (LangChain role) ───────────────────────────
print("\n" + "=" * 65)
print("         📋 STEP A: Document Summaries (LangChain Layer)")
print("=" * 65)

def simple_summary(text, num_sentences=3):
    sentences = [s.strip() for s in text.replace("\n", " ").split(".") if len(s.strip()) > 30]
    return ". ".join(sentences[:num_sentences]) + "."

summaries = {}
for i, text in enumerate(raw_texts):
    name = file_names[i].replace(".txt", "").replace("_", " ").title()
    summaries[name] = simple_summary(text)

for doc_name, summary in summaries.items():
    print(f"\n📁 Document : {doc_name}")
    print(f"📝 Summary  : {summary}")
    print("-" * 65)

# ── Step 5: Build LlamaIndex Vector Store ─────────────────────────────────────
print("\n⏳ Loading embedding model for LlamaIndex...")
Settings.embed_model = HuggingFaceEmbedding(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
Settings.llm = None  # No OpenAI key required
print("✅ Embedding model ready.")

print("\n⏳ Building vector index from legal documents...")
documents = SimpleDirectoryReader("legal_docs").load_data()
index = VectorStoreIndex.from_documents(documents)
query_engine = index.as_query_engine(similarity_top_k=2)
print("✅ Vector index built successfully.")

# ── Step 6: Legal Q&A Retrieval (LlamaIndex role) ─────────────────────────────
print("\n" + "=" * 65)
print("      ⚖️  STEP B: Legal Information Retrieval (LlamaIndex Layer)")
print("=" * 65)

legal_queries = [
    "What are the penalty clauses in the contract agreement?",
    "What was the court judgment against Carbon Industries?",
    "What are the rules about wrongful termination?",
    "What happens if an employer violates harassment policy?",
    "What environmental penalties were imposed by the court?"
]

for query in legal_queries:
    response = query_engine.query(query)
    print(f"\n⚖️  Query    : {query}")
    print(f"📌 Retrieved: {str(response).strip()}")
    print("-" * 65)

# ── Step 7: System Overview ───────────────────────────────────────────────────
print("\n" + "=" * 65)
print("   🏛️  HOW THIS SOLUTION WORKS IN PRACTICE (System Overview)")
print("=" * 65)
print("""
ARCHITECTURE:
─────────────────────────────────────────────────────────────
  Legal Documents (PDF/TXT)
        │
        ▼
  [LangChain Layer]
  • Loads and cleans raw legal documents
  • Splits large documents into smaller chunks (500 tokens)
  • Generates extractive summaries for quick review
        │
        ▼
  [LlamaIndex Layer]
  • Converts chunks into vector embeddings
  • Stores embeddings in a Vector Index (semantic search)
  • Retrieves most relevant clauses for any legal query
        │
        ▼
  Legal Professional receives:
  • Document summaries  (saves reading time)
  • Precise retrieved clauses  (answers specific questions)
─────────────────────────────────────────────────────────────

PRACTICAL BENEFITS FOR THE LEGAL FIRM:
  ✅ Summarizes 100s of pages in seconds
  ✅ Instantly retrieves relevant clauses or case details
  ✅ No OpenAI API key needed — runs fully locally
  ✅ Scales to thousands of legal documents
  ✅ Reduces manual document review time by ~80%
""")

print("✅ Legal AI System demo complete!")
```
